```python
from pyspark.sql import SparkSession
import os
import pandas as pd
# Initialize Spark session
spark = SparkSession.builder \
    .appName("Pharmacy Networks Sampling") \
    .config("spark.driver.memory", "8g") \
    .getOrCreate()
```

```python
fp = os.getcwd() 

# for i in range(1,7):
# fn = f"planinformation.txt"
# fn = os.path.join(fp, fn)

for fn in os.listdir(fp):
    if fn.endswith(".txt"):
        # Load the dataset into a PySpark DataFrame
        pharmacy_networks = spark.read.csv(
            fn, sep="|", header=True, inferSchema=True)

        # Get the original size
        original_size = pharmacy_networks.count()
        # print(f"Original size: {original_size}")

        # Randomly sample 10% of the data
        sampled_data = pharmacy_networks.sample(fraction=0.6, seed=42)


        # Get the sampled size
        sampled_size = sampled_data.count()
        print(f"Sampled size: {sampled_size}")

        # Show the first few rows of the sampled data
        sampled_data.show(5)

        output_file = os.path.join(fp, f"sampled_{fn[:-4]}.xlsx")
        sampled_data_list = sampled_data.collect()

        # Convert collected data to a DataFrame
        df = pd.DataFrame(sampled_data_list, columns=sampled_data.columns)

        # Save as Excel file
        df.to_excel(output_file, index=False)

        print(f"Sampled data saved at: {output_file}")


# output_file = os.path.join(fp, f"sampled_{fn[:-4]}.csv")
# sampled_data_list = sampled_data.collect()

# with open(output_file, "w", encoding="utf-8") as f:
#     f.write(",".join(sampled_data.columns) + "\n")  # Write headers
#     for row in sampled_data_list:
#         f.write(",".join(map(str, row)) + "\n")  # Write rows as CSV

# print(f"Sampled data saved at: {output_file}")
```
